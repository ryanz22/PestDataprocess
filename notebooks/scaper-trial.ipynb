# Scaper trial

[Scaper tutorial](https://scaper.readthedocs.io/en/latest/tutorial.html)

[Scaper git repo](https://github.com/justinsalamon/scaper)

```python
import scaper
import os
import pathlib
```

```python
path_to_audio = pathlib.Path('../data/sound/scaper_audio')

soundscape_duration = 10.0
seed = 123
foreground_folder = pathlib.Path(path_to_audio, 'foreground')
background_folder = pathlib.Path(path_to_audio, 'background')
```

```python
sc = scaper.Scaper(soundscape_duration, foreground_folder, background_folder,
                   random_state=seed)

# https://www.scienceabc.com/pure-sciences/why-negative-decibels-are-a-thing.html
sc.ref_db = -20
```

```python
sc.add_background(label=('const', 'park'),
                  source_file=('choose', []),
                  source_time=('const', 0))

sc.add_event(label=('const', 'siren'),
             source_file=('choose', []),
             source_time=('const', 0),
             #event_time=('uniform', 3, 7),
             event_time=('choose', [3, 6]),
             #event_duration=('truncnorm', 3, 1, 0.5, 5),
             event_duration=('const', 2),
             snr=('normal', 10, 3),
             #pitch_shift=('uniform', -2, 2),
             pitch_shift=None,
             #time_stretch=('uniform', 0.8, 1.2))
             time_stretch=None)

audiofile = '../soundscape.wav'
jamsfile = '../soundscape.jams'
txtfile = '../soundscape.txt'
sc.generate(audiofile, jamsfile,
            allow_repeated_label=True,
            allow_repeated_source=True,
            reverb=0.1,
            disable_sox_warnings=True,
            no_audio=False,
            txt_path=txtfile)
```
