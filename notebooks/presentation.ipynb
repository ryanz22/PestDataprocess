# Grasshopper acoustic signal processing

```python
import IPython
import librosa

DATA_FOLDER = '../data/presentation'
```

## Grasshopper sound captured by drone

NOTE: Besides drone noise, the recording also has bird call and environmental noise.

```python
orig, orig_sr = librosa.load(f'{DATA_FOLDER}/source-separate/bird-gh4-drone-mix-mono-44100.wav')
IPython.display.display(IPython.display.Audio(data=orig, rate=orig_sr))

IPython.display.Image(f'{DATA_FOLDER}/source-separate/bird-gh4-drone-mix-mono-44100.png')
```

## Source separation

The orginal sound track will be separted into 4 sources: bird, grasshopper, environmental noise and drone noise.

### Bird

```python
bird, bird_sr = librosa.load(f'{DATA_FOLDER}/source-separate/bird-gh4-drone-mix_source0-bird.wav')
IPython.display.display(IPython.display.Audio(data=bird, rate=bird_sr))

IPython.display.Image(f'{DATA_FOLDER}/source-separate/bird-gh4-drone-mix_source0-bird.png')
```

### Grasshopper

```python
gh, gh_sr = librosa.load(f'{DATA_FOLDER}/source-separate/bird-gh4-drone-mix_source1-gh4.wav')
IPython.display.display(IPython.display.Audio(data=gh, rate=gh_sr))

IPython.display.Image(f'{DATA_FOLDER}/source-separate/bird-gh4-drone-mix_source1-gh4.png')
```

### Environmental noise

```python
env, env_sr = librosa.load(f'{DATA_FOLDER}/source-separate/bird-gh4-drone-mix_source2-noise.wav')
IPython.display.display(IPython.display.Audio(data=env, rate=env_sr))

IPython.display.Image(f'{DATA_FOLDER}/source-separate/bird-gh4-drone-mix_source2-noise.png')
```

### Drone noise

```python
drone, drone_sr = librosa.load(f'{DATA_FOLDER}/source-separate/bird-gh4-drone-mix_source3-drone.wav')
IPython.display.display(IPython.display.Audio(data=drone, rate=drone_sr))

IPython.display.Image(f'{DATA_FOLDER}/source-separate/bird-gh4-drone-mix_source3-drone.png')
```

## Denoise

```python
denoise, denoise_sr = librosa.load(f'{DATA_FOLDER}/denoise/bird-gh4-drone-mix-mono-44100_denoised.wav')
IPython.display.display(IPython.display.Audio(data=denoise, rate=denoise_sr))

IPython.display.Image(f'{DATA_FOLDER}/denoise/bird-gh4-drone-mix-mono-44100_denoised.png')
```
