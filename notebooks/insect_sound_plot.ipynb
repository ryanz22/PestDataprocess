## 44100 sample rate plot

```python
fn_44k = f'{SOUND_FOLDER}/gh-4_mono_denoised.wav'
y_44k, sr_44k = librosa.load(fn_44k, sr=None)
dura_44k = librosa.get_duration(y=y_44k, sr=sr_44k)
print(f'data shape: {y_44k.shape}, data:\n{y_44k[:20]}')
print(f'sample rate: {sr_44k}')
print(f'duration: {dura_44k}')
```

```python
fig7 = plot_data(y_44k, sr=sr_44k, type='all', threshold=-60)
```

## Plot the same insect sing at different volume

### Original volume plot

```python
# Original volume
fn8 = f'{SOUND_FOLDER}volume_test/gh-4_mono_44k_denoised_0.wav'
# d8, sr8, _ = rd_file(fn8, sr=44100)
d8, sr8 = librosa.load(fn8, sr=44100)
max8 = np.max(np.abs(d8))
min8 = np.min(np.abs(d8))
mean8 = d8.mean()
# data = preprocessing.minmax_scale(data - mean, feature_range=(-1, 1))
print(f'max value of d8: {max8}, min: {min8}, mean: {mean8}')
IPython.display.Audio(data=d8, rate=sr8)
```

```python
fig8 = plot_data(d8, sr=sr8, type='all', threshold=-60)
```
