```python
import os
import numpy as np
import librosa
import librosa.display
import IPython
```

```python
SOUND_FOLDER = '../data/sound/grasshopper-sound-4/'

d1, sr1, = librosa.load(f'{SOUND_FOLDER}gh-4_mono_22050.wav', duration=5)
IPython.display.Audio(data=d1, rate=sr1)

```

```python
import ccwt

out = ccwt.fft(d1)
print(f'out:\n{out}')

freq = ccwt.frequency_band(512)
print(f'freq:\n{freq}')
```

```python
# img = ccwt.numeric_output(out, freq, 512)
# print(f'image:\n:{img}')

with open('../ccwt.png', 'wb') as f:
    ccwt.render_png(f, ccwt.RAINBOW_WALLPAPER, 10, out, freq, 512)
```
